```python
from delta.tables import DeltaTable
from pyspark.sql.types import IntegerType, StringType, TimestampType

DeltaTable.createIfNotExists(spark) \
    .tableName("Lakehouse.silver_netflix_cast") \
    .addColumn("index", IntegerType()) \
    .addColumn("person_id", IntegerType()) \
    .addColumn("id", StringType()) \
    .addColumn("name", StringType()) \
    .addColumn("character", StringType()) \
    .addColumn("role", StringType()) \
    .addColumn("load_ts", TimestampType()) \
    .execute()
```

```python
from delta.tables import DeltaTable
from pyspark.sql.functions import current_timestamp

df_credit = spark.read.format("csv").option("header","true").option("inferschema", "true").load("Files/raw_credits.csv")

target = DeltaTable.forName(spark, "Lakehouse.silver_netflix_cast")

(
    target.alias("t")
    .merge(
        df_credit.alias("s"),
        """
        t.id = s.id
        AND t.person_id = s.person_id
        AND t.role = s.role
        """
    )
    .whenMatchedUpdate(set={
        "`index`": "s.`index`",
        "name": "s.name",
        "`character`": "s.`character`",
        "load_ts": "current_timestamp()"
    })
    .whenNotMatchedInsert(values={
        "`index`": "s.`index`",
        "person_id": "s.person_id",
        "id": "s.id",
        "name": "s.name",
        "`character`": "s.`character`",
        "role": "s.role",
        "load_ts": "current_timestamp()"
    })
    .execute()
)
```

```python
df = spark.sql("SELECT * FROM Lakehouse.silver_netflix_cast LIMIT 1000")
display(df)
```

```python
from delta.tables import DeltaTable
from pyspark.sql.functions import col, current_timestamp

# Load Delta table reference

df_titles = spark.read.format("csv").option("header","true").option("inferschema","true").load("Files/raw_titles.csv")
target = DeltaTable.forName(spark, "lakehouse.silver_netflix_titles")

(
    target.alias("t")
    .merge(
        df_titles.alias("s"),
        """t.id = s.id"""   # Matching column - change if composite key needed
    )
    .whenMatchedUpdate(set={
        "index": col("s.index"),
        "title": col("s.title"),
        "type": col("s.type"),
        "release_year": col("s.release_year"),
        "age_certification": col("s.age_certification"),
        "runtime": col("s.runtime"),
        "genres": col("s.genres"),
        "production_countries": col("s.production_countries"),
        "seasons": col("s.seasons"),
        "imdb_id": col("s.imdb_id"),
        "imdb_score": col("s.imdb_score"),
        "imdb_votes": col("s.imdb_votes"),
        "update_ts": current_timestamp()
    })
    .whenNotMatchedInsert(values={
        "index": col("s.index"),
        "id": col("s.id"),
        "title": col("s.title"),
        "type": col("s.type"),
        "release_year": col("s.release_year"),
        "age_certification": col("s.age_certification"),
        "runtime": col("s.runtime"),
        "genres": col("s.genres"),
        "production_countries": col("s.production_countries"),
        "seasons": col("s.seasons"),
        "imdb_id": col("s.imdb_id"),
        "imdb_score": col("s.imdb_score"),
        "imdb_votes": col("s.imdb_votes"),
        "load_ts": current_timestamp()
    })
    .execute()
)
```
