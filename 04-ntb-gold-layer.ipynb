```sql
-- Dim Titles
CREATE TABLE IF NOT EXISTS lakehouse.gold_dim_titles
(
    title_id STRING,
    title STRING,
    type STRING,
    release_year INT,
    age_certification STRING,
    runtime INT,
    production_countries STRING,
    genres STRING
)
USING DELTA;

-- Dim Cast
CREATE TABLE IF NOT EXISTS lakehouse.gold_dim_cast
(
    person_id STRING,
    name STRING,
    role STRING,
    character STRING
)
USING DELTA;

```

```sql
-- Fact: Title-Cast Relationship
CREATE TABLE IF NOT EXISTS lakehouse.gold_fact_title_cast
(
    title_id STRING,
    person_id STRING,
    role STRING,
    character STRING
)
USING DELTA;

-- Fact: Title Metrics
CREATE TABLE IF NOT EXISTS lakehouse.gold_fact_title_metrics
(
    title_id STRING,
    imdb_score DOUBLE,
    imdb_votes BIGINT,
    total_seasons INT,
    runtime INT
)
USING DELTA;
```

```sql
MERGE INTO lakehouse.gold_dim_titles AS target
USING lakehouse.silver_netflix_titles AS source
ON target.title_id = source.id
WHEN MATCHED THEN
    UPDATE SET
        target.title = source.title,
        target.type = source.type,
        target.release_year = source.release_year,
        target.age_certification = source.age_certification,
        target.runtime = source.runtime,
        target.production_countries = source.production_countries,
        target.genres = source.genres
WHEN NOT MATCHED THEN
    INSERT (title_id, title, type, release_year, age_certification, runtime, production_countries, genres)
    VALUES (source.id, source.title, source.type, source.release_year, source.age_certification, source.runtime, source.production_countries, source.genres);
```

```sql
MERGE INTO lakehouse.gold_dim_cast AS target
USING lakehouse.silver_netflix_cast AS source
ON target.person_id = source.person_id
WHEN MATCHED THEN
    UPDATE SET
        target.name = source.name,
        target.role = source.role,
        target.character = source.character
WHEN NOT MATCHED THEN
    INSERT (person_id, name, role, character)
    VALUES (source.person_id, source.name, source.role, source.character);
```

```sql
MERGE INTO lakehouse.gold_fact_title_metrics AS target
USING lakehouse.silver_netflix_titles AS source
ON target.title_id = source.id
WHEN MATCHED THEN
    UPDATE SET
        target.imdb_score = source.imdb_score,
        target.imdb_votes = source.imdb_votes,
        target.total_seasons = source.seasons,
        target.runtime = source.runtime
WHEN NOT MATCHED THEN
    INSERT (title_id, imdb_score, imdb_votes, total_seasons, runtime)
    VALUES (source.id, source.imdb_score, source.imdb_votes, source.seasons, source.runtime);
```

```sql
MERGE INTO lakehouse.gold_fact_title_cast AS target
USING lakehouse.silver_netflix_cast AS source
ON target.title_id = source.id AND target.person_id = source.person_id
WHEN MATCHED THEN
    UPDATE SET
        target.role = source.role,
        target.character = source.character
WHEN NOT MATCHED THEN
    INSERT (title_id, person_id, role, character)
    VALUES (source.id, source.person_id, source.role, source.character);
```
